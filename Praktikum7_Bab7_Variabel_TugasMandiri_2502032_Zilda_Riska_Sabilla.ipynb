### Program 1 — Permainan Gunting, Batu, Kertas

Dua pemain memasukkan pilihan. Program membandingkan pilihan untuk menentukan pemenang dan menawarkan pengulangan permainan.

````python
pilihan = ["Gunting", "Batu", "Kertas"]
ulang = "y"

while ulang.lower() == "y":
    print("Pemain 1: Silahkan anda pilih : Gunting, Batu, Kertas")
    pemain1 = input("Pemain 1: ").capitalize()
    print("Pemain 2: Silahkan anda pilih : Gunting, Batu, Kertas")
    pemain2 = input("Pemain 2: ").capitalize()

    if pemain1 not in pilihan or pemain2 not in pilihan:
        print("Pilihan tidak tersedia, silahkan pilih Gunting, Batu, Kertas")
    elif pemain1 == pemain2:
        print("Permainan Seri")
    elif (pemain1 == "Gunting" and pemain2 == "Kertas") or \
         (pemain1 == "Batu" and pemain2 == "Gunting") or \
         (pemain1 == "Kertas" and pemain2 == "Batu"):
        print("Pemain 1 Menang")
    else:
        print("Pemain 2 Menang")

    ulang = input("Apakah Anda Ingin memulai Permainan Awal lagi [Ya/y] atau [Tidak/t] -> ")

    if ulang.lower() == "y":
        print("Permainan Baru Akan dimulai")
    else:
        print("GAME OVER")
````

**Contoh output:**

```text
Pemain 1: Silahkan anda pilih : Gunting, Batu, Kertas
Pemain 1: Gunting
Pemain 2: Silahkan anda pilih : Gunting, Batu, Kertas
Pemain 2: Kertas
Pemain 1 Menang
Apakah Anda Ingin memulai Permainan Awal lagi [Ya/y] atau [Tidak/t] -> t
GAME OVER
```

**Analisis:** Variabel `pemain1` dan `pemain2` menyimpan pilihan. Percabangan `if-elif-else` menentukan hasil berdasarkan aturan permainan. Perulangan `while` memberi kesempatan untuk mengulang permainan. Program juga memeriksa pilihan agar hanya menerima Gunting, Batu, atau Kertas.

### Program 2 — Menghitung Jumlah Huruf dan Angka

Program memeriksa setiap karakter pada teks. `isalpha()` digunakan untuk menghitung huruf, sedangkan `isdigit()` digunakan untuk menghitung angka.

````python
inputan = input("Inputan: ")
huruf = 0
angka = 0

for x in inputan:
    if x.isalpha():
        huruf += 1
    elif x.isdigit():
        angka += 1

print("Kata:", huruf)
print("Angka:", angka)
````

**Contoh output:**

```text
Inputan: password123abc
Kata: 11
Angka: 3
```

**Analisis:** Perulangan `for` membaca karakter satu per satu. Jika karakter merupakan huruf, penghitung `huruf` bertambah. Jika karakter merupakan angka, penghitung `angka` bertambah. Karakter spasi atau simbol tidak dihitung sebagai huruf maupun angka.

### Program 3 — Menampilkan Pola Angka Menurun

Program menggunakan dua perulangan `for` bersarang untuk membentuk pola angka dari 5 sampai 1.

````python
for i in range(5, 0, -1):
    for j in range(i, 0, -1):
        print(j, end=" ")
    print()
````

**Output:**

```text
5 4 3 2 1
4 3 2 1
3 2 1
2 1
1
```

**Analisis:** Perulangan luar mengatur jumlah baris dari 5 sampai 1. Perulangan dalam menampilkan angka mulai dari nilai baris saat ini hingga 1. `end=" "` membuat angka dalam satu baris tetap sejajar, sedangkan `print()` memindahkan kursor ke baris berikutnya.

### Program 4 — Menampilkan Angka Kelipatan 3 sampai 151

Program memeriksa isi list dan hanya mengambil bilangan yang habis dibagi 3 serta nilainya tidak lebih dari 151.

````python
list1 = [13, 15, 30, 41, 45, 55, 75, 95, 102, 135, 139, 151, 193, 200]
hasil = []

for angka in list1:
    if angka <= 151 and angka % 3 == 0:
        hasil.append(angka)

print("List awal:", list1)
print("Bilangan yang habis dibagi 3 dan tidak lebih dari 151:")
print(hasil)
````

**Output:**

```text
List awal: [13, 15, 30, 41, 45, 55, 75, 95, 102, 135, 139, 151, 193, 200]
Bilangan yang habis dibagi 3 dan tidak lebih dari 151:
[15, 30, 45, 75, 102, 135]
```

**Analisis:** Perulangan `for` memeriksa setiap angka dalam `list1`. Kondisi `angka <= 151` membatasi nilai maksimum, sedangkan `angka % 3 == 0` memastikan angka habis dibagi 3. Angka yang memenuhi kedua kondisi dimasukkan ke list `hasil`.

## 1. Pendahuluan
Fungsi adalah bagian program yang dibuat untuk mengerjakan tugas tertentu. Dengan fungsi, program dapat dibagi menjadi beberapa bagian yang lebih kecil sehingga susunannya lebih rapi, mudah dibaca, dan dapat digunakan kembali. Dalam Python, fungsi buatan sendiri dideklarasikan menggunakan kata kunci `def`. Fungsi dapat menerima parameter dan dapat mengembalikan hasil menggunakan `return`.

## 2. Tujuan dan Manfaat

5. Memahami penggunaan keyword argument serta variabel lokal dan global.

## 3. Dasar Teori
Bentuk umum deklarasi fungsi:

```python
def nama_fungsi(daftar_parameter):
    pernyataan
    return nilai
```

`def` digunakan untuk membuat fungsi, nama fungsi menjadi identitas untuk memanggilnya, parameter menerima data dari pemanggil, dan `return` mengirimkan hasil kembali. `return` bersifat opsional. Jika tidak ditulis, fungsi mengembalikan nilai `None`.

Fungsi tanpa parameter tidak menerima data dari pemanggil. Fungsi dengan parameter menerima satu atau beberapa nilai. Argumen dapat dikirim berdasarkan urutan (positional argument) atau nama parameter (keyword argument). Variabel lokal dibuat di dalam fungsi, sedangkan variabel global dibuat di luar fungsi.

## 4. Latihan 1 — Fungsi Tanpa Parameter dan Tanpa Nilai Kembalian
Fungsi berikut menampilkan kalimat sederhana. Fungsi tidak menerima parameter dan tidak mengembalikan nilai.

````python
# Deklarasi fungsi
def kalimat():
    """Menampilkan kalimat Hallo, Selamat Belajar Python."""
    print("Hallo, Selamat Belajar Python")

# Program utama: memanggil fungsi
kalimat()
````

**Analisis:** Fungsi `kalimat()` dideklarasikan menggunakan `def` tanpa parameter. Ketika dipanggil, perintah `print()` menampilkan teks. Karena fungsi hanya menampilkan teks dan tidak mengirimkan hasil dengan `return`, fungsi ini termasuk fungsi tanpa nilai kembalian.

## 5. Latihan 2 — Fungsi Tanpa Parameter dengan Nilai Kembalian
Fungsi berikut menjumlahkan angka 1 dan 2. Hasil perhitungan dikirim kembali menggunakan `return` lalu disimpan pada variabel `tampil`.

````python
# Deklarasi fungsi
def hitung():
    """Mengembalikan hasil penjumlahan 1 dan 2."""
    return 1 + 2

# Program utama: menampung dan menampilkan nilai kembalian
tampil = hitung()
print("Hasil Perhitungan :", tampil)
````

**Analisis:** Fungsi `hitung()` tidak memiliki parameter. Perhitungan dilakukan di dalam fungsi, kemudian `return` mengirimkan nilai 3 ke bagian program yang memanggilnya. Nilai tersebut disimpan ke variabel `tampil` dan ditampilkan dengan `print()`.

## 6. Latihan 3 — Fungsi dengan Parameter Tanpa Nilai Kembalian
Fungsi `datadiri()` menerima parameter nama dan kota. Nilai dikirim ketika fungsi dipanggil, kemudian ditampilkan dari dalam fungsi.

````python
# Deklarasi fungsi
def datadiri(nama, kota):
    """Menampilkan nama dan kota asal yang dikirim sebagai parameter."""
    print("Nama Anda :", nama)
    print("Dari Kota :", kota)

# Program utama
nama = input("Masukkan nama Anda: ")
kota = input("Masukkan kota asal: ")

# Memanggil fungsi dengan dua argumen
datadiri(nama, kota)
````

**Analisis:** Parameter `nama` dan `kota` ditulis di dalam tanda kurung pada deklarasi fungsi. Ketika `datadiri(nama, kota)` dipanggil, nilai dari variabel program utama diteruskan ke parameter sesuai urutan. Fungsi menampilkan data tersebut dan tidak menggunakan `return`.

## 7. Latihan 4 — Fungsi dengan Parameter dan Nilai Kembalian
Fungsi `perkalian()` menerima dua bilangan bulat, mengalikannya, lalu mengembalikan hasil perkalian.

````python
# Deklarasi fungsi
def perkalian(bil1, bil2):
    """Menghitung perkalian dua bilangan."""
    hasil = bil1 * bil2
    return hasil

# Program utama
bil1 = int(input("Input bil 1: "))
bil2 = int(input("Input bil 2: "))

# Memanggil fungsi dan menampilkan nilai kembaliannya
print("Hasil Perkalian", bil1, "dengan", bil2, "adalah", perkalian(bil1, bil2))
````

**Analisis:** Parameter `bil1` dan `bil2` menerima dua bilangan dari pengguna. Variabel `hasil` menyimpan hasil perkalian. Pernyataan `return hasil` mengirimkan hasil ke pemanggil fungsi. Contohnya, masukan 5 dan 7 menghasilkan 35.

## 8. Latihan 5 — Melewatkan Argumen dengan Kata Kunci (Keyword Argument)
Keyword argument mengirim nilai dengan menyebutkan nama parameter. Dengan cara ini, urutan penulisan argumen dapat diubah selama setiap parameter diberi nilai satu kali.

````python
def perkalian(bil1, bil2):
    """Menghitung perkalian dua bilangan."""
    hasil = bil1 * bil2
    return hasil

# Contoh pemanggilan dengan keyword argument dalam urutan terbalik
hasil = perkalian(bil2=7, bil1=5)
print("Hasil perkalian dengan keyword argument:", hasil)
````

**Analisis:** Pada pemanggilan `perkalian(bil2=7, bil1=5)`, nilai 7 diberikan kepada parameter `bil2` dan nilai 5 diberikan kepada `bil1`, meskipun urutannya berbeda dari deklarasi. Nama parameter harus tepat dan tidak boleh diberi nilai dua kali dalam satu pemanggilan.

## 9. Latihan 6 — Variabel Lokal dan Variabel Global
Variabel global didefinisikan di luar fungsi sehingga dapat dibaca di bagian program lain. Variabel lokal didefinisikan di dalam fungsi dan hanya berlaku pada ruang lingkup fungsi tersebut.

````python
# Variabel global
pesan = "Belajar fungsi Python"

def contoh_lokal():
    # Variabel lokal
    nilai = 90
    print("Pesan global dari dalam fungsi:", pesan)
    print("Nilai lokal di dalam fungsi:", nilai)

# Program utama
contoh_lokal()
print("Pesan global dari program utama:", pesan)

# Variabel nilai tidak dipanggil di luar fungsi karena bersifat lokal
````

**Analisis:** Variabel `pesan` dibuat di luar fungsi sehingga dapat dibaca dari dalam fungsi dan program utama. Variabel `nilai` dibuat di dalam `contoh_lokal()`, sehingga hanya dapat digunakan di dalam fungsi tersebut. Jika `nilai` dipanggil dari luar fungsi, Python akan menghasilkan `NameError` karena nama tersebut tidak dikenal pada ruang lingkup program utama.

## 10. Kesimpulan
Berdasarkan latihan, fungsi pada Python dibuat menggunakan `def` dan dipanggil dengan menuliskan nama fungsi beserta argumennya. Fungsi dapat dibuat tanpa parameter maupun dengan parameter. Fungsi juga dapat mengembalikan nilai menggunakan `return` atau hanya menjalankan perintah tanpa mengembalikan nilai secara eksplisit. Keyword argument memudahkan pengiriman nilai berdasarkan nama parameter. Selain itu, variabel lokal hanya berlaku di dalam fungsi, sedangkan variabel global didefinisikan di luar fungsi. Penggunaan fungsi membuat program lebih terstruktur, mudah dibaca, dan dapat digunakan kembali.

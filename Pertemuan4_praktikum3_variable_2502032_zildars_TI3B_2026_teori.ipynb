## 2. Pengertian Variabel
Variabel adalah nama yang digunakan untuk menyimpan suatu nilai. Nilai tersebut dapat digunakan kembali atau diubah sesuai kebutuhan.

Contoh:
- `nama` menyimpan teks.
- `umur` menyimpan angka.
- `hobi` menyimpan teks.

Python bersifat *case-sensitive*, artinya `nama` dan `Nama` dianggap sebagai dua variabel yang berbeda.


## 3. Contoh Variabel Sederhana

```python
nama = "Song Kang"
umur = 32
aktor_favorit = "Song Kang"
aktris_favorit = "Go Youn-jung"

print(nama)
print(umur)
print(aktor_favorit)
print(aktris_favorit)
```

**Analisis:** `nama`, `umur`, `aktor_favorit`, dan `aktris_favorit` adalah variabel. Teks ditulis di antara tanda petik, sedangkan angka dapat ditulis tanpa tanda petik. Perintah `print()` digunakan untuk menampilkan isi variabel.

## 4. Latihan 1 — Membuat Program Biodata

Program ini menggunakan variabel untuk menyimpan biodata. Nama diisi sesuai identitas, sedangkan alamat, umur, dan hobi dimasukkan saat program dijalankan. Nama aktor dan aktris favorit menjadi contoh tambahan variabel bertipe teks.

```python
# Program biodata
deskripsi = "Program biodata menggunakan variabel Python"
nama = "Zilda Riska Sabilla"
alamat = input("Masukkan alamat: ")
umur = int(input("Masukkan umur: "))
hobi = input("Masukkan hobi: ")

aktor_favorit = "Song Kang"
aktris_favorit = "Go Youn-jung"

print("Nama saya", nama, "yang beralamat di", alamat,
      "umur sekarang", umur, "tahun dan memiliki hobi", hobi)
print("Aktor favorit:", aktor_favorit)
print("Aktris favorit:", aktris_favorit)
```

## 5. Latihan 2 — Menghitung Luas dan Keliling Persegi Panjang

Diketahui panjang = 15 cm dan lebar = 5.7 cm.

Rumus:
- Luas = panjang × lebar
- Keliling = 2 × (panjang + lebar)


```python
# Menghitung luas dan keliling persegi panjang
panjang = 15
lebar = 5.7

luas_persegi_panjang = panjang * lebar
keliling_persegi_panjang = 2 * (panjang + lebar)

print("Sebuah persegi panjang memiliki panjang", panjang,
      "cm, lebar", lebar, "cm dan memiliki luas sebesar",
      luas_persegi_panjang, "cm persegi dan keliling",
      keliling_persegi_panjang, "cm")
```

**Analisis program:**
- Variabel `panjang` berisi nilai 15.
- Variabel `lebar` berisi nilai 5.7.
- Perkalian panjang dan lebar menghasilkan luas 85.5 cm persegi.
- Dua kali jumlah panjang dan lebar menghasilkan keliling 41.4 cm.

## 4. Soal 1 — Menampilkan Karakter Berdasarkan Indeks

Program disusun mengikuti instruksi pada soal kuis.

````python
kalimat = "PRAKTIKUM ALPRO1"
print("String seluruhnya:", kalimat)
print("Menampilkan hanya 1 huruf pertama dari string, yaitu", kalimat[0])
print("Menampilkan huruf terakhir dari string yaitu", kalimat[-1])
print("Menampilkan huruf ke-2, ke-3, ke-5, dan ke-14:")
print("Huruf ke-2:", kalimat[1])
print("Huruf ke-3:", kalimat[2])
print("Huruf ke-5:", kalimat[4])
print("Huruf ke-14:", kalimat[14])
````

**Contoh keluaran:**

```text
Huruf ke-2: R
Huruf ke-3: A
Huruf ke-5: T

```

**Analisis:** Indeks Python dimulai dari 0, sehingga karakter ke-2 diambil dari indeks 1. Karakter terakhir dapat diambil dengan indeks -1. Spasi juga dihitung sebagai karakter.

## 5. Soal 2 — Swapping Number

Program disusun mengikuti instruksi pada soal kuis.

````python
x = 20
y = 10
print("nilai awal x= 20, y= 10")
temp = x
x = y
y = temp
print("nilai x setelah di tukar:", x)
print("nilai y setelah di tukar:", y)
````

**Contoh keluaran:**

```text
nilai awal x= 20, y= 10
nilai x setelah di tukar: 10
nilai y setelah di tukar: 20
```

**Analisis:** Nilai x disimpan ke temp, kemudian x diisi nilai y dan y diisi nilai temp. Hasil akhirnya, nilai kedua variabel berhasil bertukar.

## 6. Soal 3 — Operator Bitwise

Program disusun mengikuti instruksi pada soal kuis.

````python
a = 20
b = 9
print("Nilai dari operasi bitwise AND antara a:20, b:9 -> (a and b) =", a & b)
print("Nilai dari operasi bitwise OR antara a:20, b:9 -> (a or b) =", a | b)
print("Nilai dari operasi bitwise XOR antara a:20, b:9 -> (a xor b) =", a ^ b)
````

**Contoh keluaran:**

```text
Nilai dari operasi bitwise AND antara a:20, b:9 -> (a and b) = 0
Nilai dari operasi bitwise OR antara a:20, b:9 -> (a or b) = 29
Nilai dari operasi bitwise XOR antara a:20, b:9 -> (a xor b) = 29
```

**Analisis:** Operator &, |, dan ^ digunakan untuk operasi bitwise AND, OR, dan XOR. Pada bilangan 20 dan 9, hasil AND adalah 0, sedangkan OR dan XOR menghasilkan 29.

## 7. Soal 4 — Perhitungan Midpoint

Program disusun mengikuti instruksi pada soal kuis.

````python
print("Program Kalkulasi midpoint")
x1, y1 = 2, 2
x2, y2 = 4, 4
midpoint_x = (x1 + x2) / 2
midpoint_y = (y1 + y2) / 2
print("\nNilai midpoint dari garis:")
print("Nilai midpoint x :", midpoint_x)
print("Nilai midpoint y :", midpoint_y)
````

**Contoh keluaran:**

```text
Program Kalkulasi midpoint

Nilai midpoint dari garis:
Nilai midpoint x : 3.0
Nilai midpoint y : 3.0
```

**Analisis:** Midpoint dihitung dengan merata-ratakan koordinat x dan y. Titik tengah dari (2, 2) dan (4, 4) adalah (3.0, 3.0).

## 8. Soal 5 — Menjumlahkan Digit Bilangan

Program disusun mengikuti instruksi pada soal kuis.

````python
angka = 1234
jumlah = 0
sisa = angka
while sisa > 0:
    digit = sisa % 10
    jumlah += digit
    sisa //= 10
print("angka =", angka)
print("Jumlah dari angka =", jumlah)
````

**Contoh keluaran:**

```text
angka = 1234
Jumlah dari angka = 10
```

**Analisis:** Modulo 10 mengambil digit terakhir, sedangkan pembagian bulat 10 menghilangkan digit terakhir. Proses diulang sampai bilangan habis. Hasil 1 + 2 + 3 + 4 adalah 10.

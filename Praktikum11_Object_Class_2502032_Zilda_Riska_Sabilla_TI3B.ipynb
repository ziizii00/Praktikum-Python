## 3. Program dan Analisis

### 3.1 Contoh class Employee

Contoh berikut memperlihatkan class dengan class variable, constructor, instance variable, dan method. Data karyawan disimpan pada object yang dibuat dari class Employee.

````python
class Employee:
    empCount = 0

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.empCount += 1

    def displayEmployee(self):
        print("Name:", self.name, ", Salary:", self.salary)

emp1 = Employee("Zara", 2000)
emp2 = Employee("Manni", 5000)

emp1.displayEmployee()
emp2.displayEmployee()
print("Total Employee", Employee.empCount)
````

**Output:**

```text
Name: Zara , Salary: 2000
Name: Manni , Salary: 5000
Total Employee 2
```

**Analisis:** `Employee` merupakan class. `emp1` dan `emp2` adalah object. Constructor `__init__()` dijalankan saat object dibuat untuk mengisi atribut `name` dan `salary`. `empCount` merupakan class variable yang nilainya digunakan bersama oleh seluruh object, sedangkan `displayEmployee()` merupakan method untuk menampilkan data karyawan.

### 3.2 Program A — Detail Personal

Class `Person` memiliki class variable `person` dan dua instance variable, yaitu `rambut` dan `warna`. Dua object dibuat untuk menyimpan detail Budi dan Michael.

````python
# Class untuk menyimpan detail seseorang
class Person:
    # Class Variable
    person = "Person"

    # Constructor
    def __init__(self, rambut, warna):
        # Instance Variable
        self.rambut = rambut
        self.warna = warna

# Membuat object
Budi = Person("Ikal", "Hitam")
Michael = Person("Lurus", "Cokelat")

print("Budi details:")
print("Rambut:", Budi.rambut)
print("Warna Rambut:", Budi.warna)

print("\nMichael details:")
print("Rambut:", Michael.rambut)
print("Warna Rambut:", Michael.warna)
````

**Output:**

```text
Budi details:
Rambut: Ikal
Warna Rambut: Hitam

Michael details:
Rambut: Lurus
Warna Rambut: Cokelat
```

**Analisis:** `Person` adalah class yang menjadi cetak biru object. Method `__init__()` merupakan constructor yang menerima dua parameter dan menyimpannya ke atribut object melalui `self`. Object `Budi` dan `Michael` memiliki nilai atribut masing-masing. Nilai rambut dan warna Michael digunakan sebagai contoh pengisian object.

## 4. Mengubah Program Modul Praktikum 10 menjadi Object & Class

Empat contoh berikut mengubah pola fungsi yang dipelajari sebelumnya menjadi method di dalam class. Setiap program membuat object terlebih dahulu, kemudian memanggil method melalui object tersebut.

### Program 1 — Method tanpa parameter dan tanpa nilai kembali

````python
class Pesan:
    def tampilkan(self):
        print("Hallo, Selamat Belajar Python")

# Membuat object dan memanggil method
objek = Pesan()
objek.tampilkan()
````

**Output:**

```text
Hallo, Selamat Belajar Python
```

**Analisis:** Class `Pesan` memiliki method `tampilkan()` tanpa parameter tambahan dan tidak mengembalikan nilai. Method dipanggil melalui object `objek`.

### Program 2 — Method tanpa parameter dengan nilai kembali

````python
class Perhitungan:
    def hitung(self):
        return 1 + 2

# Membuat object dan menampilkan nilai kembalian method
objek = Perhitungan()
tampil = objek.hitung()
print("Hasil Perhitungan:", tampil)
````

**Output:**

```text
Hasil Perhitungan: 3
```

**Analisis:** Method `hitung()` melakukan penjumlahan dan mengembalikan hasil dengan `return`. Nilai tersebut disimpan ke variabel `tampil` lalu dicetak.

### Program 3 — Method dengan parameter tanpa nilai kembali

````python
class DataDiri:
    def tampilkan(self, nama, kota):
        print("Nama Anda:", nama)
        print("Dari Kota:", kota)

# Membuat object
objek = DataDiri()

# Data yang diberikan sebagai argumen
nama = "Zilda"
kota = "Tegal"
objek.tampilkan(nama, kota)
````

**Output:**

```text
Nama Anda: Zilda
Dari Kota: Tegal
```

**Analisis:** Method `tampilkan()` menerima parameter `nama` dan `kota`. Data diteruskan saat method dipanggil. Method menampilkan data dan tidak menggunakan `return`.

### Program 4 — Method dengan parameter dan nilai kembali

````python
class Perkalian:
    def hitung(self, bil1, bil2):
        hasil = bil1 * bil2
        return hasil

# Membuat object
objek = Perkalian()

bil1 = 5
bil2 = 7
print("Hasil Perkalian", bil1, "dengan", bil2, "adalah", objek.hitung(bil1, bil2))
````

**Output:**

```text
Hasil Perkalian 5 dengan 7 adalah 35
```

**Analisis:** Method `hitung()` menerima dua parameter bilangan, mengalikan keduanya, kemudian mengembalikan hasil perkalian. Hasil ditampilkan melalui `print()`.

### 3.1 Percobaan Tipe Data: Boolean dan String

Boolean memiliki dua nilai, yaitu `True` dan `False`. String digunakan untuk menyimpan teks. Fungsi `type()` dapat digunakan untuk memeriksa tipe data suatu nilai.

```python
# Tipe data Boolean
nilai_boolean = False
print(nilai_boolean)
print(type(nilai_boolean))

# Tipe data String
teks = "Ini adalah tulisan berupa String"
print(teks)
print(type(teks))
```

### 3.2 Percobaan Escape Sequence

Escape sequence adalah kombinasi karakter yang diawali dengan garis miring terbalik (`\\`) untuk memberi arti khusus di dalam string. Pada contoh `\b` dan `\r`, tampilan dapat bergantung pada cara terminal merender karakter kontrol. `repr()` digunakan agar bentuk karakter khusus lebih mudah diperiksa.

```python
# Escape sequence: petik tunggal dan petik ganda
print("Petik tunggal: \'")
print("Petik ganda: \"")

# \n membuat baris baru
print("Baris pertama\nBaris kedua")

# \\ menampilkan satu karakter garis miring terbalik
print("Folder: C:\\Python")

# \b adalah backspace; repr memperlihatkan karakter kontrolnya
print(repr("ABC\bD"))

# \t membuat tab
print("Nama\tNilai")
print("V\t100")

# \r adalah carriage return; repr memperlihatkan karakter kontrolnya
print(repr("Awal\rAkhir"))
```

### 3.3 Percobaan Tipe Data Numerik

Integer menyimpan bilangan bulat, Float menyimpan bilangan pecahan/desimal, sedangkan Binary, Octal, dan Hexadecimal merupakan penulisan bilangan dengan basis 2, 8, dan 16. Awalan Python masing-masing adalah `0b`, `0o`, dan `0x`.

```python
# Integer
angka_bulat = 100
print(angka_bulat)
print(type(angka_bulat))

# Float
angka_desimal = 0.001
print(angka_desimal)
print(type(angka_desimal))

# Binary (basis 2), Octal (basis 8), dan Hexadecimal (basis 16)
bilangan_biner = 0b1010
bilangan_oktal = 0o12
bilangan_heksadesimal = 0x0A

print(bilangan_biner)
print(bilangan_oktal)
print(bilangan_heksadesimal)

# Contoh konversi heksadesimal ke bilangan desimal
print("bilangan desimal dari 0x01 adalah", 0x01)
```

### 3.4 Percobaan Tipe Data Complex

Tipe data `complex` digunakan untuk bilangan kompleks. Dalam Python, bagian imajiner ditandai dengan huruf `j`.

```python
# Tipe data Complex
bilangan_kompleks = 2 + 6j
print(type(bilangan_kompleks))
print(bilangan_kompleks)
print(type(2 + 6j))
```

### 3.5 Percobaan List dan Tuple

List ditulis menggunakan tanda kurung siku `[]` dan dapat diubah setelah dibuat. Tuple ditulis menggunakan tanda kurung biasa `()` dan umumnya digunakan untuk kumpulan data yang tidak perlu diubah.

```python
# Tipe data List
daftar_angka = [96, 97, 98, 99, 100]
daftar_kata = ['seratus', 'dua ratus', 'tiga ratus']
print(daftar_angka)
print(daftar_kata)
print(type(daftar_angka))

# Tipe data Tuple
tuple_angka = (96, 97, 98, 99, 100)
tuple_kata = ('seratus', 'dua ratus', 'tiga ratus')
print(tuple_angka)
print(tuple_kata)
print(type(tuple_angka))
```

### 3.6 Percobaan Dictionary dan Fungsi `type()`

Dictionary menyimpan data sebagai pasangan kunci dan nilai. Pada contoh berikut, dictionary berisi nama dan umur. Fungsi `type()` mengembalikan tipe objek, misalnya `<class 'dict'>` untuk dictionary.

```python
# Dictionary langsung
biodata_contoh = {'nama': 'Ani', 'umur': 19}
print(biodata_contoh)
print(type(biodata_contoh))

# Variabel yang menyimpan dictionary
biodata_idol = {'nama': 'Kim Taehyung', 'umur': 30}
print(biodata_idol)
print(type(biodata_idol))
```

### 3.7 Program Latihan Nomor 2 — Output Wajib

Kode berikut disusun agar menghasilkan teks dan nilai yang diminta pada instruksi. Nilai `Ani`, umur `19`, dan isi list dipertahankan persis.

```python
# Program latihan Praktikum Algoritma dan Pemrograman 1
print("Praktikum Algoritma dan Pemrograman 1 - Variable dan Tipe Data")
print(False)
print(type(False))
print("Ini adalah tulisan berupa String")
print(100)
print(0.001)
print("bilangan desimal dari 0x01 adalah", 0x01)
print(10)
print(type(2 + 6j))
print(2 + 6j)
print(type(2 + 6j))
print([96, 97, 98, 99, 100])
print(['seratus', 'dua ratus', 'tiga ratus'])
print({'nama': 'Ani', 'umur': 19})
print("This string contains a single quote (') character.")
```

### 3.8 Program Boolean — `boolean.py`

Program ini mengikuti kode yang diberikan pada modul. Variabel `narkoba` bernilai `False` dan `belajar` bernilai `True`.

```python
# Tipe Data Boolean
# Nama file jika disimpan sebagai script: boolean.py
narkoba = False
belajar = True

print(narkoba)
print(belajar)
```

**Penjelasan:** Variabel adalah nama yang merujuk pada nilai yang disimpan. `narkoba` dan `belajar` bertipe Boolean karena nilainya `False` dan `True`. Fungsi `print()` menampilkan nilai variabel ke output. Hasilnya adalah `False` pada baris pertama dan `True` pada baris kedua.

### 3.9 Program String Format — `string_format.py`

Contoh berikut menggunakan nama Kim Taehyung (V), sesuai izin mengganti nama contoh pada bagian ini.

```python
# String Format
# Nama file jika disimpan sebagai script: string_format.py
first_name = "Kim"
middle_name = "Taehyung"
last_name = "V"

# f-string memasukkan nilai variabel ke dalam teks
sapa = f"Halo {first_name} {middle_name} {last_name}"

print(sapa)
```

**Penjelasan:** F-string adalah string yang diawali huruf `f`. Ekspresi atau nama variabel yang ditulis di dalam kurung kurawal `{}` akan diganti dengan nilai variabel tersebut. Karena itu, output program adalah `Halo Kim Taehyung V`.

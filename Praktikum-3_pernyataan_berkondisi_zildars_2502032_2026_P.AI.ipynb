## 3.1 Menentukan Bilangan Terbesar
Program membandingkan dua bilangan dan menampilkan bilangan yang lebih besar.

```python
a = int(input("Masukkan bilangan pertama: "))
b = int(input("Masukkan bilangan kedua: "))

if a > b:
    print("Bilangan terbesar:", a)
else:
    print("Bilangan terbesar:", b)
```

## 3.2 Menentukan Jenis Akar Persamaan Kuadrat
Diskriminan dihitung menggunakan rumus `D = b**2 - 4*a*c`.

```python
a = float(input("Masukkan nilai a: "))
b = float(input("Masukkan nilai b: "))
c = float(input("Masukkan nilai c: "))

D = b**2 - 4*a*c

if D < 0:
    print("Akar Imaginer")
elif D > 0:
    print("Akar Nyata Berlainan")
else:
    print("Akar Kembar")
```

## 3.3 Operator Logika
Operator `and` mengharuskan kedua kondisi benar, `or` memerlukan minimal satu kondisi benar, dan `not` membalik nilai logika.

```python
x = 5

print(x > 0 and x < 10)
print(x < 0 or x == 5)
print(not (x > 10))
```

## 3.3 Pengelompokan Umur
Program mengelompokkan umur sesuai rentang kategori pada modul.

```python
umur = int(input("Masukkan umur: "))

if umur <= 5:
    print("Balita")
elif umur <= 13:
    print("Anak-anak")
elif umur <= 25:
    print("Remaja")
elif umur <= 35:
    print("Dewasa")
elif umur <= 55:
    print("Orang Tua")
else:
    print("Lansia")
```

## 3.4 Pernyataan Berkondisi Bersarang
Contoh ini memperlihatkan `if` di dalam blok `if` lainnya.

```python
x = 8
y = 12

if x > 5:
    if y > 10:
        print("x lebih besar dari 5 dan y lebih besar dari 10")
    else:
        print("x lebih besar dari 5 dan y tidak lebih besar dari 10")
else:
    print("x tidak lebih besar dari 5")
```

## 3.5 Latihan — Pengelompokan Nilai
Masukkan nilai untuk mendapatkan kriteria A sampai E sesuai batas pada halaman latihan.

```python
nilai = int(input("Masukkan nilai: "))

if nilai >= 88:
    print("Kriteria A")
elif nilai >= 77:
    print("Kriteria B")
elif nilai >= 60:
    print("Kriteria C")
elif nilai >= 45:
    print("Kriteria D")
else:
    print("Kriteria E")
```

## 2. Materi Singkat

Perulangan digunakan ketika ada perintah yang perlu dijalankan berkali-kali. Dengan `for`, kita tidak perlu menulis perintah yang sama satu per satu.

Bentuk dasarnya:

```python
for variabel in iterable:
    perintah
```

`range()` dapat digunakan dalam tiga bentuk:
- `range(nilai_akhir)` dimulai dari 0.
- `range(nilai_awal, nilai_akhir)` dimulai dari nilai awal.
- `range(nilai_awal, nilai_akhir, pencacah)` menggunakan jarak kenaikan tertentu.

Nilai akhir pada `range()` tidak ikut ditampilkan, jadi jika ingin sampai angka 10, batas akhirnya ditulis 11.

## 3. Contoh Penggunaan `range()`

### 3.1 Menampilkan bilangan 1 sampai 10 dengan pencacah 1

````python
for i in range(1, 11):
    print(i, end=" ")
````

**Penjelasan:** Program menampilkan angka 1 sampai 10. Nilai `i` berubah satu angka setiap perulangan.

### 3.2 Menampilkan bilangan 1 sampai 10 dengan pencacah 2

````python
for i in range(1, 11, 2):
    print(i, end=" ")
````

**Penjelasan:** Pencacahnya adalah 2, sehingga angka yang tampil memiliki selisih dua, yaitu 1, 3, 5, 7, dan 9.

### 3.3 Menampilkan bilangan dari nilai awal sampai nilai akhir

````python
awal = 3
akhir = 12

for i in range(awal, akhir + 1):
    print(i, end=" ")

print()
print("Banyak bilangan:", len(range(awal, akhir + 1)))
print("Jumlah seluruh bilangan:", sum(range(awal, akhir + 1)))
````

**Penjelasan:** Nilai awal dan akhir disimpan dalam variabel. `akhir + 1` digunakan supaya angka akhir ikut ditampilkan. Program juga menghitung banyak bilangan dan jumlah seluruhnya.

### 3.4 Menampilkan bilangan dari nol sampai batas tertentu

````python
batas = 10

for i in range(batas + 1):
    print(i, end=" ")
````

**Penjelasan:** Jika hanya satu nilai yang diberikan pada `range()`, perulangan dimulai dari 0. Batas ditambah 1 agar angka 10 ikut muncul.

## 4. Latihan 1 – Menampilkan Bilangan dengan Pencacah 3 dan Menghitung Perkalian
Latihan ini mengikuti perintah pada bagian `range(nilai_awal, nilai_akhir, pencacah)` di modul. Program menampilkan bilangan dari nilai awal sampai nilai akhir dengan kenaikan 3, kemudian menghitung hasil perkalian semua bilangan yang tampil.

Contoh data yang digunakan: awal = 1 dan akhir = 10.

````python
awal = 1
akhir = 10
hasil_kali = 1

print("Bilangan yang ditampilkan:")
for i in range(awal, akhir + 1, 3):
    print(i, end=" ")
    hasil_kali *= i

print("\nHasil perkalian semua bilangan:", hasil_kali)
````

**Penjelasan:** Perulangan dimulai dari 1 dan bertambah 3 setiap putaran. Angka yang diproses adalah 1, 4, 7, dan 10. Variabel `hasil_kali` diawali dengan 1, lalu dikalikan dengan setiap angka yang ditemukan.

## 5. Perulangan `for` Bertingkat
Perulangan bertingkat adalah perulangan yang berada di dalam perulangan lain. Biasanya digunakan untuk menampilkan data dalam bentuk baris dan kolom.

### 5.1 Contoh menampilkan baris dan kolom

````python
for baris in range(1, 4):
    for kolom in range(1, 5):
        print(baris * kolom, end="\t")
    print()
````

**Penjelasan:** Perulangan luar mengatur baris, sedangkan perulangan dalam mengatur kolom. Setiap angka baris dikalikan dengan angka kolom.

## 6. Latihan 2 – Penjumlahan Dua Buah Matriks
Latihan ini mengikuti bagian latihan perulangan bertingkat pada modul. Dua matriks dijumlahkan pada posisi baris dan kolom yang sama.

````python
matriks_a = [
    [1, 2, 3],
    [4, 5, 6]
]

matriks_b = [
    [7, 8, 9],
    [1, 2, 3]
]

baris = len(matriks_a)
kolom = len(matriks_a[0])

hasil = []

for i in range(baris):
    baris_hasil = []
    for j in range(kolom):
        jumlah = matriks_a[i][j] + matriks_b[i][j]
        baris_hasil.append(jumlah)
    hasil.append(baris_hasil)

print("Matriks A:")
for data in matriks_a:
    print(data)

print("\nMatriks B:")
for data in matriks_b:
    print(data)

print("\nHasil penjumlahan:")
for data in hasil:
    print(data)
````

**Penjelasan:** Program mengambil elemen matriks A dan B pada posisi yang sama, kemudian menjumlahkannya. Perulangan luar digunakan untuk baris dan perulangan dalam untuk kolom. Hasil setiap baris disimpan ke dalam matriks `hasil`.

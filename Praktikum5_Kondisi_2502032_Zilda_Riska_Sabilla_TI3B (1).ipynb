## 3. Dasar Teori: Kondisi `if`
Kondisi `if` menjalankan blok kode ketika ekspresi bernilai `True`. Jika bernilai `False`, blok tersebut dilewati. Kondisi `if-else` menyediakan dua pilihan, sedangkan `if-elif-else` digunakan ketika terdapat beberapa kemungkinan.

Operator yang digunakan pada latihan ini antara lain `==` (sama dengan), `>` dan `<` (perbandingan), `or` (salah satu kondisi benar), `and` (semua kondisi benar), serta `%` (sisa bagi).

### Contoh sederhana `if` dan `if-else`

```python
nilai = 9

if nilai > 7:
    print("Selamat, Anda lulus")

nilai = 3
if nilai > 7:
    print("Selamat, Anda lulus")
else:
    print("Maaf, Anda tidak lulus")
```

**Analisis:** Pada contoh pertama, nilai 9 lebih besar dari 7 sehingga perintah pada blok `if` dijalankan. Pada contoh kedua, nilai 3 tidak lebih besar dari 7 sehingga perintah pada blok `else` dijalankan.

## 4. Program Latihan 1 — Menentukan Jenis Segitiga
Program meminta tiga panjang sisi. Sebelum menentukan jenisnya, program memeriksa apakah setiap sisi positif dan apakah ketiga sisi memenuhi syarat pembentukan segitiga. Setelah itu, kondisi digunakan untuk membedakan segitiga sama sisi, sama kaki, dan sembarang.

```python
# Program 1 - Menentukan jenis segitiga
print("PROGRAM MENENTUKAN JENIS SEGITIGA")

x = float(input("Masukkan panjang sisi pertama: "))
y = float(input("Masukkan panjang sisi kedua: "))
z = float(input("Masukkan panjang sisi ketiga: "))

if x <= 0 or y <= 0 or z <= 0:
    print("Panjang sisi harus lebih dari 0.")
elif x + y <= z or x + z <= y or y + z <= x:
    print("Ketiga sisi tersebut tidak dapat membentuk segitiga.")
elif x == y and y == z:
    print("Jenis segitiga: Segitiga Sama Sisi (Equilateral Triangle)")
elif x == y or x == z or y == z:
    print("Jenis segitiga: Segitiga Sama Kaki (Isosceles Triangle)")
else:
    print("Jenis segitiga: Segitiga Sembarang (Scalene Triangle)")
```

**Contoh pengujian:** masukkan 5, 5, dan 5 untuk memperoleh segitiga sama sisi. Masukkan 5, 5, dan 7 untuk memperoleh segitiga sama kaki. Masukkan 4, 5, dan 6 untuk memperoleh segitiga sembarang.

**Analisis:** Kondisi pertama menolak sisi yang tidak positif. Kondisi berikutnya memeriksa ketaksamaan segitiga. Jika ketiga sisi sama, program memilih sama sisi. Jika hanya dua sisi yang sama, program memilih sama kaki. Selain itu, program menampilkan segitiga sembarang.

## 5. Program Latihan 2 — Menentukan Tahun Kabisat
Tahun kabisat memiliki 366 hari. Aturan pemeriksaannya adalah tahun yang habis dibagi 400 merupakan kabisat; tahun yang habis dibagi 100 tetapi tidak 400 bukan kabisat; selain itu, tahun yang habis dibagi 4 merupakan kabisat.

```python
# Program 2 - Menentukan tahun kabisat
print("PROGRAM MENENTUKAN TAHUN KABISAT")

tahun = int(input("Masukkan tahun: "))

if tahun <= 0:
    print("Tahun harus berupa bilangan positif.")
elif tahun % 400 == 0:
    print(tahun, "merupakan tahun kabisat.")
elif tahun % 100 == 0:
    print(tahun, "bukan tahun kabisat.")
elif tahun % 4 == 0:
    print(tahun, "merupakan tahun kabisat.")
else:
    print(tahun, "bukan tahun kabisat.")
```

**Contoh pengujian:** tahun 2020 merupakan tahun kabisat, sedangkan tahun 2019 bukan tahun kabisat. Tahun 1900 bukan kabisat karena habis dibagi 100 tetapi tidak habis dibagi 400. Tahun 2000 merupakan kabisat.

**Analisis:** Operator `%` digunakan untuk memperoleh sisa pembagian. Jika sisanya 0, berarti tahun habis dibagi angka tersebut. Urutan `if` dan `elif` penting agar aturan khusus untuk kelipatan 100 dan 400 diperiksa dengan benar.

## 6. Program Latihan 3 — Menentukan Zodiak Yunani
Program meminta tanggal dan bulan lahir, kemudian mencocokkannya dengan rentang tanggal zodiak. Batas tanggal yang digunakan mengikuti rentang umum zodiak.

```python
# Program 3 - Menentukan zodiak Yunani berdasarkan tanggal lahir
print("PROGRAM MENENTUKAN ZODIAK")

tanggal = int(input("Masukkan tanggal lahir (1-31): "))
bulan = int(input("Masukkan bulan lahir (1-12): "))

if bulan < 1 or bulan > 12 or tanggal < 1:
    print("Tanggal atau bulan tidak valid.")
elif (bulan == 1 and tanggal <= 19) or (bulan == 12 and tanggal >= 22):
    print("Zodiak Anda adalah Capricorn")
elif (bulan == 1 and tanggal >= 20) or (bulan == 2 and tanggal <= 18):
    print("Zodiak Anda adalah Aquarius")
elif (bulan == 2 and tanggal >= 19) or (bulan == 3 and tanggal <= 20):
    print("Zodiak Anda adalah Pisces")
elif (bulan == 3 and tanggal >= 21) or (bulan == 4 and tanggal <= 19):
    print("Zodiak Anda adalah Aries")
elif (bulan == 4 and tanggal >= 20) or (bulan == 5 and tanggal <= 20):
    print("Zodiak Anda adalah Taurus")
elif (bulan == 5 and tanggal >= 21) or (bulan == 6 and tanggal <= 20):
    print("Zodiak Anda adalah Gemini")
elif (bulan == 6 and tanggal >= 21) or (bulan == 7 and tanggal <= 22):
    print("Zodiak Anda adalah Cancer")
elif (bulan == 7 and tanggal >= 23) or (bulan == 8 and tanggal <= 22):
    print("Zodiak Anda adalah Leo")
elif (bulan == 8 and tanggal >= 23) or (bulan == 9 and tanggal <= 22):
    print("Zodiak Anda adalah Virgo")
elif (bulan == 9 and tanggal >= 23) or (bulan == 10 and tanggal <= 22):
    print("Zodiak Anda adalah Libra")
elif (bulan == 10 and tanggal >= 23) or (bulan == 11 and tanggal <= 21):
    print("Zodiak Anda adalah Scorpio")
elif (bulan == 11 and tanggal >= 22) or (bulan == 12 and tanggal <= 21):
    print("Zodiak Anda adalah Sagittarius")
else:
    print("Tanggal tidak sesuai dengan jumlah hari pada bulan tersebut.")
```

**Contoh pengujian:** tanggal 15 bulan 8 menghasilkan Leo, sedangkan tanggal 10 bulan 1 menghasilkan Capricorn.

**Analisis:** Setiap zodiak memiliki rentang tanggal yang diperiksa menggunakan gabungan kondisi `and` dan `or`. Program memeriksa rentang bulan dan tanggal lalu menampilkan nama zodiak yang sesuai. Pemeriksaan ini merupakan penyederhanaan berdasarkan tanggal-bulan umum; validasi jumlah hari pada setiap bulan dapat dikembangkan lebih lanjut.

## 7. Program Latihan 4 — Menentukan Shio
Program menentukan shio berdasarkan tahun lahir menggunakan siklus 12 tahunan astrologi Tionghoa. Daftar shio disimpan dalam list, kemudian indeks ditentukan menggunakan operasi modulo.

```python
# Program 4 - Menentukan shio berdasarkan tahun lahir
print("PROGRAM MENENTUKAN SHIO")

tahun_lahir = int(input("Masukkan tahun lahir: "))

daftar_shio = [
    "Tikus", "Kerbau", "Macan", "Kelinci",
    "Naga", "Ular", "Kuda", "Kambing",
    "Monyet", "Ayam", "Anjing", "Babi"
]

if tahun_lahir <= 0:
    print("Tahun lahir harus berupa bilangan positif.")
else:
    indeks_shio = (tahun_lahir - 4) % 12
    shio = daftar_shio[indeks_shio]
    print("Tahun lahir:", tahun_lahir)
    print("Shio Anda adalah:", shio)
```

**Contoh pengujian:** tahun 2020 menghasilkan shio Tikus, tahun 2021 Kerbau, dan tahun 2024 Naga.

**Analisis:** Rumus `(tahun_lahir - 4) % 12` menghasilkan indeks dari 0 sampai 11. Indeks tersebut digunakan untuk mengambil nama shio dari `daftar_shio`. Cara ini memanfaatkan pola berulang setiap 12 tahun.
